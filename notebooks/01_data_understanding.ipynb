```python
print("Total rows:", len(sales_10k))

print("Missing promo_id:", sales_10k["promo_id"].isna().sum())

print("Non-missing promo_id:", sales_10k["promo_id"].notna().sum())

print("\nUnique promo IDs:")
print(sales_10k["promo_id"].dropna().nunique())
```

```python
# Duplicate rows remove

print("Before removing duplicates:", len(sales_10k))

sales_10k = sales_10k.drop_duplicates().copy()

print("After removing duplicates:", len(sales_10k))
print("Duplicates removed:", 10000 - len(sales_10k))
```

```python
numeric_columns = [
    "quantity",
    "unit_price",
    "total_value",
    "discount_pct"
]

print(sales_10k[numeric_columns].describe())
```

```python
for col in ["quantity", "unit_price", "total_value", "discount_pct"]:
    print(col, "negative values:", (sales_10k[col] < 0).sum())
```

```python
for col in ["quantity", "unit_price", "total_value"]:
    print(col, "zero values:", (sales_10k[col] == 0).sum())
```

```python


calculated_value = (
    sales_10k["quantity"]
    * sales_10k["unit_price"]
    * (1 - sales_10k["discount_pct"] / 100)
)

difference = (sales_10k["total_value"] - calculated_value).abs()

print("Maximum difference:", difference.max())
print("Rows with difference > 0.01:", (difference > 0.01).sum())
```

```python
print("CHANNEL VALUES:")
print(sales_10k["channel"].value_counts())

print("\nUnique Stores:", sales_10k["store_id"].nunique())
print("Unique SKUs:", sales_10k["sku_id"].nunique())
print("Unique Customers:", sales_10k["customer_id"].nunique())
print("Unique Receipts:", sales_10k["receipt_id"].nunique())
```

```python
# Extracting from  Original data 
sales_final = pd.read_csv(
    "../data/retail_clean_dataset/sales_transactions.csv",
    nrows=10050
)

# Duplicate rows remove
sales_final = sales_final.drop_duplicates().copy()

# Date format
sales_final["date"] = pd.to_datetime(
    sales_final["date"],
    errors="coerce"
)

# Exactly 10,000 records
sales_final = sales_final.head(10000).copy()

print("Final records:", len(sales_final))
print("Final columns:", sales_final.shape[1])
print("Duplicates:", sales_final.duplicated().sum())
print("Missing values:", sales_final.isnull().sum().sum())
```

```python
sales_final.to_csv(
    "../data/sales_10k_clean.csv",
    index=False
)

print("Clean dataset saved successfully!")
```
