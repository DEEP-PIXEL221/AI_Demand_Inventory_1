```python
import pandas as pd
import os

data_path = "../data/retail_clean_dataset"

sales = pd.read_csv(
    "../data/sales_10k_clean.csv"
)

sku_master = pd.read_csv(
    os.path.join(data_path, "sku_master.csv")
)

store_master = pd.read_csv(
    os.path.join(data_path, "store_master.csv")
)

customer_master = pd.read_csv(
    os.path.join(data_path, "customer_master.csv")
)

promotions = pd.read_csv(
    os.path.join(data_path, "promotions.csv")
)

inventory = pd.read_csv(
    os.path.join(data_path, "inventory_snapshot.csv")
)

print("Sales:", sales.shape)
print("SKU Master:", sku_master.shape)
print("Store Master:", store_master.shape)
print("Customer Master:", customer_master.shape)
print("Promotions:", promotions.shape)
print("Inventory:", inventory.shape)
```

```python
datasets = {
    "sales": sales,
    "sku_master": sku_master,
    "store_master": store_master,
    "customer_master": customer_master,
    "promotions": promotions,
    "inventory": inventory
}

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name.upper())
    print(df.columns.tolist())
```

```python
print("Sales records:", len(sales))

print("Unique receipt_id:", sales["receipt_id"].nunique())
print("Unique store_id:", sales["store_id"].nunique())
print("Unique sku_id:", sales["sku_id"].nunique())
print("Unique customer_id:", sales["customer_id"].nunique())
print("Unique promo_id:", sales["promo_id"].nunique(dropna=True))
```

```python
sales_skus = set(sales["sku_id"].dropna().unique())
master_skus = set(sku_master["sku_id"].dropna().unique())

missing_skus = sales_skus - master_skus

print("Unique SKUs in sales:", len(sales_skus))
print("Unique SKUs in SKU master:", len(master_skus))
print("SKUs missing from master:", len(missing_skus))

if len(missing_skus) > 0:
    print("Missing SKU IDs:", list(missing_skus)[:20])
```

```python
#store matching check
sales_stores = set(sales["store_id"].dropna().unique())
master_stores = set(store_master["store_id"].dropna().unique())

missing_stores = sales_stores - master_stores

print("Unique stores in sales:", len(sales_stores))
print("Unique stores in store master:", len(master_stores))
print("Stores missing from master:", len(missing_stores))

if len(missing_stores) > 0:
    print("Missing Store IDs:", list(missing_stores)[:20])
```

```python
#customer matching check 
sales_customers = set(sales["customer_id"].dropna().unique())
master_customers = set(customer_master["cust_id"].dropna().unique())

missing_customers = sales_customers - master_customers

print("Unique customers in sales:", len(sales_customers))
print("Unique customers in customer master:", len(master_customers))
print("Customers missing from master:", len(missing_customers))

if len(missing_customers) > 0:
    print("Missing Customer IDs:", list(missing_customers)[:20])
```
